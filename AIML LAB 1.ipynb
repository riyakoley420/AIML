### Problem 1: BFS and DFS Traversal of a Graph



```python
# Define the graph using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

print("Graph defined:")
for node, neighbors in graph.items():
    print(f"{node}: {neighbors}")
```

```python
from collections import deque

def bfs(graph, start_node):
    visited = []
    queue = deque([start_node])

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.append(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return visited

# Perform BFS traversal starting from node A
bfs_traversal_order = bfs(graph, 'A')
print(f"BFS Traversal Order: {bfs_traversal_order}")
```

```python
def dfs(graph, start_node):
    visited = []
    stack = [start_node] # Using a list as a stack

    while stack:
        node = stack.pop()
        if node not in visited:
            visited.append(node)
            # Add neighbors to stack in reverse order to explore in alphabetical order (if applicable)
            # or simply to maintain consistent exploration if order matters for visualization
            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return visited

# Perform DFS traversal starting from node A
dfs_traversal_order = dfs(graph, 'A')
print(f"DFS Traversal Order: {dfs_traversal_order}")
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Function to draw the graph with traversal order
def draw_traversal_tree(graph, traversal_order, title):
    G = nx.DiGraph(graph) # Use directed graph for tree structure representation

    pos = nx.spring_layout(G, seed=42) # For consistent layout

    plt.figure(figsize=(10, 7))
    plt.title(title, size=15)

    # Draw nodes
    nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=2000)

    # Draw edges
    nx.draw_networkx_edges(G, pos, edge_color='gray', arrows=True, arrowsize=20)

    # Add labels with traversal order
    labels = {}
    for i, node in enumerate(traversal_order):
        labels[node] = f'{node} ({i+1})'
    nx.draw_networkx_labels(G, pos, labels, font_size=12)

    plt.axis('off')
    plt.show()

# Draw BFS Traversal Tree
draw_traversal_tree(graph, bfs_traversal_order, 'BFS Traversal Order Tree')

# Draw DFS Traversal Tree
draw_traversal_tree(graph, dfs_traversal_order, 'DFS Traversal Order Tree')
```
